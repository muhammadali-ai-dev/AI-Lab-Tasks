```python
import heapq
class search:
    def __init__(self):
        self.graph = {
            "s": [("a",1), ("b",4)],
            "a": [("c",2)],
            "b": [("g",5)],
            "c": [("g",3)],
            "g": []
        }
        self.h = {
            "s":5, "a":4, "b":4,
            "c":2,"g":0
        }
    def astar(self,start,goal):
            q = [(self.h[start], 0, start, [start])]
            best = {start:0}
            while q:
                f, cost, node, path = heapq.heappop(q)
                if cost != best[node]:
                    continue
                if node == goal:
                    return path, cost
                for n, w in self.graph[node]:
                    new_cost = cost + w
                    if new_cost < best.get(n, float("inf")):
                        best[n] = new_cost
                        heapq.heappush(q, (new_cost + self.h[n], new_cost, n, path + [n]))
    def ucs(self, start, goal):
        q = [(0, start, [start])]
        best = {start: 0}
        expanded = []
        while q:
            cost, node, path = heapq.heappop(q)
            if cost != best[node]:
                continue
            if node == goal:
                return path, cost, expanded
            expanded.append(node)
            for n, w in self.graph[node]:
                new_cost = cost + w
                if new_cost < best.get(n, float("inf")):
                    best[n] = new_cost
                    heapq.heappush(q, (new_cost, n, path + [n]))
s = search()
print(s.astar("s","g"))
print(s.ucs("s","g"))
s.h["c"] = 10
print("overestimated:", s.astar("s","g"))
```
