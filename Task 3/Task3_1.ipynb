```python
from collections import deque
class search:
    def __init__(self):
        self.graph = {
            "a": ["b","c"],
            "b": ["d","e"],
            "c": ["f"],
            "d": [],
            "e": ["g"],
            "f": [],
            "g":[]
        }
    def bfs(self, start, goal):
        q = deque([[start]])
        while q:
            path = q.popleft()
            node = path[-1]
            if node == goal:
                return path
            for n in self.graph[node]:
                q.append(path + [n])
    def dfs(self, start, goal):
        stack = [[start]]
        while stack:
            path = stack.pop()
            node = path[-1]
            if node == goal:
                return path
            for n in reversed(self.graph[node]):
                if n not in path:
                    stack.append(path + [n])
s = search()
print(s.bfs("a","g"))
print(s.dfs("a","g"))

```
