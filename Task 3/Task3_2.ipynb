```python
class search:
    def __init__(self):
        self.graph = {
            "a": ["b","c"],
            "b": ["d","e"],
            "c": ["f"],
            "d": [],
            "e": ["g"],
            "f": [],
            "g":[]
        }
    def dls(self, node, goal, limit, path):
        if node == goal:
            return path
        if limit == 0:
            return None
        for n in self.graph[node]:
            if n not in path:
                result = self.dls(n, goal, limit -1, path +[n])
                if result:
                    return result
        return None
    def iddfs(self,start,goal,maximum_depth):
        for limit in range(maximum_depth + 1):
            result = self.dls(start, goal, limit,[start])
            if result:
                return result
        return None
s = search()
print(s.dls("a","g",3, ["a"]))
print(s.iddfs("a","g",4))        
            
```
