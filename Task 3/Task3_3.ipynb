```python
import heapq
class search:
    def __init__(self):
       self.graph = {
        "a": [("b",1),("c",6)],
        "b": [("d",2),("e",1)],
        "c":[("f",2),("g",6)],
        "d":[],
        "e":[("g",1)],
        "f":[],
        "g":[]
    }
    def ucs(self, start, goal):
        q = [(0, start, [start])]
        visited = {}
        while q:
            cost, node, path =heapq.heappop(q)
            if node == goal:
                return path, cost
            if node in visited and visited[node] <= cost:
                continue
            visited[node] = cost
            for n, weight in self.graph[node]:
                heapq.heappush(q,(cost + weight, n, path + [n]))
s = search()
print(s.ucs("a","g"))    
```
