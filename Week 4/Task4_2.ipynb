```python
import heapq
class search:
    def __init__(self):
        self.graph = {
            "s": [("a",1), ("b",4)],
            "a": [("c",2)],
            "b": [("g",5)],
            "c": [("g",3)],
            "g": []
        }
        self.h = {
            "s":5, "a":4, "b":4,
            "c":2,"g":0
        }
    def astar(self,start,goal):
            q = [(self.h[start], 0, start, [start])]
            best = {start:0}
            expanded = []
            while q:
                f, cost, node, path = heapq.heappop(q)
                if cost != best[node]:
                    continue
                if node == goal:
                    return path, cost, expanded
                expanded.append(node)
                    
                for n, w in self.graph[node]:
                    new_cost = cost + w
                    if new_cost < best.get(n, float("inf")):
                        best[n] = new_cost
                        heapq.heappush(q, (new_cost + self.h[n], new_cost, n, path + [n]))
    def ucs(self, start, goal):
        q = [(0, start, [start])]
        best = {start: 0}
        expanded = []
        while q:
            cost, node, path = heapq.heappop(q)
            if cost != best[node]:
                continue
            if node == goal:
                return path, cost, expanded
            expanded.append(node)
            for n, w in self.graph[node]:
                new_cost = cost + w
                if new_cost < best.get(n, float("inf")):
                    best[n] = new_cost
                    heapq.heappush(q, (new_cost, n, path + [n]))
def path_cost(graph, path):
    total = 0
    for i in range(len(path)-1):
        for n, w in graph[path[i]]:
            if n == path[i + 1]:
                total = total + w
    return total
   
```

```python
s = search()
a = s.astar("s", "g")
print("A star search")
print("path:", a[0])
print("Returned cost:", a[1])
print("Actual path cost:", path_cost(s.graph,a[0]))
print("processed nodes:", a[2])
print("Expansion count:", len(a[2]))
u = s.ucs("s", "g")
print("\nUCS sEARCh")
print("path:", u[0])
print("Returned cost:", u[1])
print("Actual path cost:", path_cost(s.graph,u[0]))
print("processed nodes:", u[2])
print("Expansion count:", len(u[2]))
```
