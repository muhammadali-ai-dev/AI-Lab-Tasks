```python
class Rule_system:
    def __init__(self,score,documents,attendance,interview_score):
        self.score = score
        self.documents = documents
        self.attendance = attendance
        self.interview_score = interview_score
    def check_result(self):
        if (self.score >= 70 and
            self.documents == "complete" and 
            self.attendance >=75 and
            self.interview_score >=60):
            return "pass"
        else:
            return"fail"
#case 1
case1 = Rule_system(80,"complete",70,75)
print("case 1:",case1.check_result())
```

```python
#case 2
case2 = Rule_system(80,"complete",85,55)
print("case 2:",case2.check_result())
```

```python
#case 3
case3 = Rule_system(70,"complete",75,60)
print("case 3:",case3.check_result())
```
